```python
import numpy as np
from scipy.spatial import distance
pointA = np.array([2,4,6])
pointB = np.array([5,1,9])

#Euclidean Distance
euclidean_dist = distance.euclidean(pointA,pointB)
print("Euclidean Distance: ", euclidean_dist)

#Similarity(inverse of distance)
euclidean_similarity = 1/(1+euclidean_dist)
print("\nEuclidean Similarity: ", euclidean_similarity)

#Manhattan Distance
manhattan_dist = distance.cityblock(pointA,pointB)
print("\nManhattan Distance: ", manhattan_dist)

#Manhattan Similarity
manhattan_similarity = 1/(1+manhattan_dist)
print("\nManhattan Similarity: ", manhattan_similarity)

#Minkowski Distance
minkowski_dist = distance.minkowski(pointA,pointB)
print("\nMinkowski Distance: ", minkowski_dist)

#Minkowski Similarity
minkowski_similarity = 1/(1+minkowski_dist)
print("\nMinkowski Similarity: ", minkowski_similarity)
```

```python
import pandas as pd
from scipy.stats import spearmanr

df = pd.DataFrame ({
    'X': [10,20,30,40,50],
    'Y': [12,24,33,45,60]
})

#Spearman correlation coefficient and p-value
corr_value, p_value = spearmanr(df['X'],df['Y'])
print(f"Spearman Correlation Coefficient: ", corr_value)
print(f"p-value: ", p_value)
```

```python
def hamming_distance(str1, str2):
  if len(str1) != len(str2):
    raise ValueError("Strings must be of equal length")

  return sum(ch1 != ch2 for ch1, ch2 in zip(str1, str2))

s1 = "karolin"
s2 = "kathrin"
dist = hamming_distance(s1, s2)
print(f"Hamming Distance between '{s1}' and '{s2}': {dist}")
```

```python
def hamming_distance(str1, str2):
  if len(str1) != len(str2):
    raise ValueError("Strings must be of equal length")

  return sum(ch1 != ch2 for ch1, ch2 in zip(str1, str2))

s1 = input()
s2 = input()
dist = hamming_distance(s1, s2)
print(f"Hamming Distance between '{s1}' and '{s2}': {dist}")
```

```python
def jaccard_index(str1, str2):
  set1, set2 = set(str1.split()), set(str2.split())
  intersection = set1.intersection(set2)
  union = set1.union(set2)
  return len(intersection) / len(union)

s1 = input()
s2 = input()
print("Jaccard Index:",jaccard_index(s1,s2))
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
s1 = input()
s2 = input()
vectorizer = CountVectorizer().fit([s1,s2])
vectors = vectorizer.transform([s1,s2])
cos_sim = cosine_similarity(vectors[0], vectors[1])[0][0]

print("Cosine Similarity:",cos_sim)
```

```python
def lcs_length(X, Y):
  m, n = len(X), len(Y)
  dp = [[0]*(n+1) for _ in range(m+1)]
  for i in range(m):
    for j in range(n):
      if X[i] == Y[j]:
        dp[i+1][j+1] = dp[i][j]+1
      else:
        dp[i+1][j+1] = max(dp[i][j+1], dp[i+1][j])
  return dp[m][n]
seq1 = "ABCDEF"
seq2 = "AEBDF"
length = lcs_length(seq1, seq2)
print(f"Longest Common Subsequence length between '{seq1}' and '{seq2}': {length}")
```
